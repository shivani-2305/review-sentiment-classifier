```python
from google import genai
from google.colab import userdata
client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))
def classify_review(review_text):
     prompt=f"""
Classify the review as Positive,Negative or Mixed. Reply with only one word
Review:"I loved the ambience and the food was good"
Sentiment:Positive
Review:"Worst experience and food was bad"
Sentiment:Negative
Review:"Food was good but service was slow"
Sentiment:Mixed
Review:"{review_text}"
Sentiment:
"""
     response=client.models.generate_content(model="gemini-3.8-flash",contents=prompt,)
     return response.text
review="the people are good,but this man was worst"
result = classify_review(review)
print("Review:",review)
print("result:",result)
```

```python
from google import genai
from google.colab import userdata
client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))
MODEL="gemini-3.8-flash"
def summarize_text(long_text):
    prompt=f"""
summarize the following text in 2-3 simple senteces.
text:"{long_text}"
summary:
"""
    response = client.models.generate_content(model=MODEL,contents=prompt,)
    return response.text.strip()

text="""Python is a popular programming language created by guido van rossum in 1991.It is known for its simple,readable system which makes it a great choice for beginners.python is widely used in web development,data structures,artificial intelligence,and automation.many top companies like netflix,google,andinstagram use python in the systems."""
result = summarize_text(text)
print("original length", len(text))
print("summary", result)
```

```python
!pip install -q -U google-genai
from google import genai
from google.genai import types
from google.colab import userdata
client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))
MODEL="gemini-3.5-flash-lite"
chat = client.chats.create(model=MODEL,config=types.GenerateContentConfig(system_instruction="tell me about how to plan for trips for badrinath.explain in simple words with 5 examples"),)
print("chatbot ready")
while True:
   user_input = input("you:")
   if user_input.lower() in ["exit", "quit"]:
         print("bye")
         break
   reply = chat.send_message(user_input)
   print("Bot:", reply.text)
```
